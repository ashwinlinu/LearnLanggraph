```python
from langgraph.graph import StateGraph
from typing import TypedDict, Optional
```

```python
class compliment_agent(TypedDict):
    name: str
    message: Optional[str] = None

def compliment_node(state: compliment_agent) -> compliment_agent:
    current_message = state.get('message', "")
    state['message'] = f"{current_message} {state['name']} you are doing great job learning langGraph"
    return state
```

```python
graph = StateGraph(compliment_agent)

graph.add_node("compliment_agent", compliment_node)
graph.set_entry_point("compliment_agent")
graph.set_finish_point("compliment_agent")
```

```python
app = graph.compile()
result = app.invoke({"name": "Ashwin"})
```
