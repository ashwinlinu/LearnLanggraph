```python
from langgraph.graph import StateGraph
from typing import TypedDict, List, Literal
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    operation: Literal["*", "+"]
    result : str
```

```python
def calulator_agent(state: AgentState) -> AgentState:
    operation = state['operation']
    values = state['values']
    if operation == "*":
        result = 1
        for value in values:
            result *= value
    elif operation == "+":
        result = sum(values)
    else:
        raise ValueError(f"Invalid operation: {operation}")
    state['result'] = f"Hi {state['name']}, your sum is {result}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("calculator_agent", calulator_agent)

graph.set_entry_point("calculator_agent")
graph.set_finish_point("calculator_agent")

app = graph.compile()
result = app.invoke({"name": "Ashwin", "values": [1,2,3,4,5], "operation": "+"})
```
