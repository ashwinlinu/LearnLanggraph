```python
from langgraph.graph import StateGraph
from typing import TypedDict, List
```

```python
class AgentState(TypedDict):
    values : List[int]
    name : str
    result : str
```

```python
def calculate_sum(state: AgentState) -> AgentState:
    """ this node will calculate the sum of the values in the state"""

    state["result"] = f"{state.get('name', 'user')}, your sum is {sum(state['values'])}"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("input_calculator", calculate_sum)

graph.set_entry_point("input_calculator")
graph.set_finish_point("input_calculator")

app = graph.compile()
result = app.invoke({"values": [1, 2, 3, 4, 5], "name" : "Ashwin"})
```
