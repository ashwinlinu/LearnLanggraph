```python
from typing import TypedDict, Dict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name : str
    message : str

def greating_node(state : AgentState) -> AgentState:
    """
    Simple node that add Greating Message to State
    """
    state['message'] = f"Hello {state['name']}, how are you?"
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("greeter", greating_node)
graph.set_entry_point("greeter")
graph.set_finish_point("greeter")
```

```python
app = graph.compile()
```

```python
result = app.invoke({"name": "Ashwin", "message": "Hello"})
```
